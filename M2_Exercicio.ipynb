**Exercício: Gerenciador de Tarefas**

Crie um programa em Python que funcione como um gerenciador de tarefas.
O programa deve permitir ao usuário:

- Adicionar uma nova tarefa.
- Listar todas as tarefas.
- Remover uma tarefa pelo número.
- Sair do programa.

Utilize um loop para manter o programa em execução até que o usuário escolha sair.

```python
tarefas = []




def adicionar_tarefa():
    tarefa = input("digite a tarefa que deseja adicionar")
    tarefas.append(tarefa)

    print(f"Tarefa {tarefa} adicionada com sucesso!")

def listar_tarefa():
    if not tarefas:
        print("não há tarefas")
    else:
        for i in range(len(tarefas)):
            print(f"{i+1}. {tarefas[i]}")


def remover_tarefa():
    if not tarefas:
        print("tarefa inexistente")
    
    else:
        
        for i in range(len(tarefas)):
            print(f"{i}. {tarefas[i]}")
    
            indice = int(input("Digite o numero da tarefa que deseja remover"))
    
        if(0<= indice < len(tarefas)):
            remover = tarefas.pop(indice)
            print(f"tarefa {remover} removida com sucesso")
        else:
            print("tarefa invalida")




while True:
    print("\n=====Lista de tarefas=====")
    print("\n1. Adicionar uma nova tarefa")
    print("\n2. Listar tarefas")
    print("\n3. Remover uma tarefa pelo  seu numero")
    print("\n4. Sair")

    opcao = input("Selecione uma opção:")

    match opcao:

        case "1":  
          adicionar_tarefa()

        case "2":
            listar_tarefa()

        case "3":
            remover_tarefa()


        case "4":
            print("saindo...")
            break

        case _:
            print("opção invalida")


                
            





```
